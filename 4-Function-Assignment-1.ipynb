```python
def oddEven():
    num=int(input("Enter a number"))
    if(num%2==0):
        print(num, "is even number")
        message="even"
    else:
        print(num, " is odd number")
        message="odd"
    return message
```

```python
message=oddEven()
```

```python
print(message)
```

```python
def marriage():
    age=int(input("enter your age"))
    gender=input("enter your gender")
    if(age<18 and gender=="female"):
        print("Your Gender:",gender)
        print("Your Age:",age)
        print("Not eligible")
        message="Not eligible"
    if(age>=18 and gender=="female"):
        print("Your Gender:",gender)
        print("Your Age:",age)
        print("eligible")
        message="eligible"
    if(age<21 and gender=="male"):
        print("Your Gender:",gender)
        print("Your Age:",age)
        print("Not eligible")
        message="Not eligible"
    if(age>=21 and gender=="male"):
        print("Your Gender:",gender)
        print("Your Age:",age)
        print("eligible")
        message="eligible"
    return message
```

```python
eligible=marriage()
```

```python
def percentage(sub1,sub2,sub3,sub4,sub5):
    percent=(sub1+sub2+sub3+sub4+sub5)/5
    return percent
```

```python
percentage(98,87,95,95,93)
```

```python
def triangeArea(height,breadth):
    area=(height*breadth)/2
    print("Area of Triangle:",area)
    return area
```

```python
def triangePerimeter(height1,height2,breadth):
    perimeter=height1+height2+breadth
    print("Perimeter of Triangle:",perimeter)
    return perimeter
```

```python
triangleArea=triangeArea(2,1)
print(triangleArea)
```

```python
trianglePerimeter=triangePerimeter(2,2,1)
print(trianglePerimeter)
```
